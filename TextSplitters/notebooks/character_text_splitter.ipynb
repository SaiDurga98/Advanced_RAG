```python
from langchain_text_splitters import CharacterTextSplitter

from pprint import pp
```

```python
text = """Artificial intelligence (AI) is transforming the world at an unprecedented pace. Machine learning, a subset of AI, enables computers to learn from data without being explicitly programmed. Deep learning takes this further by using neural networks with many layers to recognize patterns in complex data.

Natural Language Processing (NLP) allows machines to understand and generate human language. Applications include chatbots, translation services, and sentiment analysis. Large Language Models (LLMs) like GPT and Claude are trained on massive datasets and can perform a wide range of language tasks.

Retrieval-Augmented Generation (RAG) is a technique that combines information retrieval with text generation. Instead of relying solely on a model's training data, RAG fetches relevant documents from a knowledge base and uses them to generate more accurate and up-to-date responses. This makes RAG particularly useful for question-answering systems.

Vector databases store embeddings — numerical representations of text. When a user asks a question, the query is converted into an embedding and compared against stored embeddings to find the most relevant documents. Popular vector databases include Pinecone, Chroma, and FAISS.

LangChain is a framework that simplifies building applications powered by language models. It provides tools for document loading, text splitting, embedding, vector storage, and chaining together multiple AI components into a pipeline."""
```

```python
# Split the text into chunks
chunks = text_splitter.split_text(text)

pp(chunks)
```

```python
from termcolor import COLORS, colored
from random import choice
colors_list = list(COLORS.keys())[2:8]
colors_list
```

```python
def print_colored_chunks(chunks):
    colors_list = list(COLORS.keys())[2:8]
    print(f"Total Number of Chunks: {len(chunks)}")
    for num, chunk in enumerate(chunks, 1):
        print(f"Chunk {num}: Length {len(chunk)} chars")
        print(colored(text=chunk, color=choice(colors_list)), end="\n\n")
```

```python
print_colored_chunks(chunks)
```

```python
# Create a function to create chunks based on the charcter separator : cuts the word in the middle if the chunk size is less than the word length
def create_chunks(text:str, chunk_size: int, separator: str, chunk_overlap:int = 0) -> list[str]:
    text_splitter = CharacterTextSplitter(chunk_size=chunk_size,chunk_overlap=chunk_overlap,separator=separator)
    chunks = text_splitter.split_text(text)
    return chunks
```

```python
print_colored_chunks(create_chunks(text, chunk_size=100, separator="", chunk_overlap=0))
```

```python
# Increase the chunk size to 400 characters and print the colored chunks again
print_colored_chunks(create_chunks(text, chunk_size=400, separator="", chunk_overlap=0))
```

```python
# lets use word separator instead of character and see the difference : words doesn't get cut in the middle
print_colored_chunks(create_chunks(text, chunk_size=100, separator=" ", chunk_overlap=0))
```

```python
# Chunking based on paragraphs using double new line as separator
print_colored_chunks(create_chunks(text, chunk_size=100, separator="\n\n", chunk_overlap=0))
```

```python
# Create a function with chunk overlap of 10 characters

def display_chunks_with_overlap(chunks, chunk_overlap):
    if not chunk_overlap:
        raise ValueError("Chunk overlap must be greater than 0")
    
    overlap_text = ""
    overlap_color = "white"
    colors_list = list(COLORS.keys())[2:8]
    print(f"Total Number of Chunks: {len(chunks)}")
    for num, chunk in enumerate(chunks, 1):
        print(f"Chunk {num}: Length {len(chunk)} chars")
        if num == 1:
            non_overlap_text = chunk[0:-chunk_overlap]
            overlap_text = chunk[-chunk_overlap:]
            print(colored(text=non_overlap_text, color=choice(colors_list)), colored(text=overlap_text, color=overlap_color), sep="" ,end="\n\n")
        else:
            print(colored(text=overlap_text, color=overlap_color), end="")
            non_overlap_text = chunk[chunk_overlap:-chunk_overlap]
            print(colored(text=non_overlap_text, color=choice(colors_list)), end="")
            overlap_text = chunk[-chunk_overlap:]
            print(colored(text=overlap_text, color=overlap_color), end="\n\n")

```

```python
display_chunks_with_overlap(chunks = create_chunks(text, chunk_size=300, separator=" "),chunk_overlap=30)
```

### For splitting documents we can use split_documents
