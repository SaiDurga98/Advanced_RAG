```python
from langchain_community.document_loaders.web_base import WebBaseLoader
from pathlib import Path
from pprint import pp
```

```python
# WebBaseLoader is a document loader that can be used to load documents from the web. It takes a list of URLs as input and returns a list of Document objects. 
# Each Document object contains the content of the web page and its metadata, such as the URL and the title of the page.
# BeautifulSoup is used to parse the HTML content of the web page and extract the relevant information. 
# The WebBaseLoader can be used to load documents from any website, as long as the content can be accessed and parsed using BeautifulSoup.

url = "https://docs.langchain.com/oss/python/integrations/document_loaders/web_base"
web_loader = WebBaseLoader(web_path=url)
documents = web_loader.load()
print(documents[0].page_content)
pp(documents[0].metadata)
```
