```python
from langchain_community.document_loaders.csv_loader import CSVLoader
from pathlib import Path
from pprint import pp
```

```python
file_path = Path("../knowledge-source/organizations.csv")
```

```python
#csv_loader = CSVLoader(file_path)
csv_loader = CSVLoader(file_path,source_column="Industry",content_columns=["Description"])
documents = csv_loader.load()
print(documents[0])
pp(documents[0].metadata)

# source_column is used to specify which column in the csv file should be used as the source of the document. This will be stored in the metadata of the document under the key "source".
# content_columns is used to specify which columns in the csv file should be used as the content of the document. If not specified, the entire row will be used as the content. If multiple columns are specified, their values will be concatenated to form the content of the document.
```
