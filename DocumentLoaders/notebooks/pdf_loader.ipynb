```python
from langchain_community.document_loaders.pdf import PyPDFLoader
from pathlib import Path
from pprint import pp
```

```python
file_path = Path("../knowledge-source/attention_is_all_you_need.pdf")
```

```python
pypdf_loader = PyPDFLoader(file_path,mode='page')
```

```python
documents = pypdf_loader.load()
```

```python
pp(documents[2].page_content)
```

```python
from langchain_community.document_loaders.parsers import TesseractBlobParser, RapidOCRBlobParser
```

```python
py_pdf_immage_loader = PyPDFLoader(file_path,mode='page',extract_images=True,images_parser=TesseractBlobParser())
```

```python
documents_with_images = py_pdf_immage_loader.load()
```

```python
page_with_image = documents_with_images[2]
```

```python
print(page_with_image.page_content)
```

```python
from langchain_community.document_loaders.pdf import PDFMinerLoader
pdf_miner_loader = PDFMinerLoader(file_path,mode='page',extract_images=True,images_parser=RapidOCRBlobParser(),images_inner_format="html-img")
pdf_miner_documents = pdf_miner_loader.load()
print(pdf_miner_documents[2].page_content)
```

```python
from langchain_community.document_loaders.pdf import PDFPlumberLoader # detailed metadata, but no images support yet
pdf_plumber_loader = PDFPlumberLoader(file_path)
pdf_plumber_documents = pdf_plumber_loader.load()
pp(pdf_plumber_documents[2].metadata)
```
