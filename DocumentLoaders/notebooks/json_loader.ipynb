```python
def metadata_func(record:dict, metadata:dict):
    metadata["product_id"] = record["productID"]
    metadata["product_name"] = record["productName"]

    # delete seq number from metadata as it is not useful for retrieval and will only add noise to the metadata
    del metadata["seq_num"]
    return metadata
```

```python
from langchain_community.document_loaders.json_loader import JSONLoader
from pathlib import Path
from pprint import pp
file_path = Path("../knowledge-source/apparels.json")
#json_loader = JSONLoader(file_path, jq_schema=".products[]",text_content=False)
#json_loader = JSONLoader(file_path, jq_schema=".products[].Description")
json_loader = JSONLoader(file_path, jq_schema=".products[]",content_key="Description",metadata_func=metadata_func)

# content_key is used to specify which key in the json record should be used as the content of the document. If not specified, the entire json record will be used as the content.
# metadata_func is used to specify a function that takes the json record and the metadata dictionary as input and returns the modified metadata dictionary. This can be used to add custom metadata fields or modify existing ones.
```

```python
documents = json_loader.load()
```

```python
pp(documents)
print(documents[0].metadata)
```
